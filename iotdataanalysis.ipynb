```python
#Import Libraries
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
#Load iot data
df = pd.read_csv("iot_sensor_data.csv")
df["Timestamp"] = pd.to_datetime(df["Timestamp"])
print("IoT Sensor Data\n",df)
```

```python
#Statical Analysis
temp = df["Temperature"].values
mean_temp = np.mean(temp) #mean(avg)
std_temp = np.std(temp) #standard deviation
min_temp = np.min(temp) # minimum temp
max_temp = np.max(temp) #max temp
print("\n Mean Temperature :",mean_temp)
print("Standard Deviation Temperature :",std_temp)
print("\n Minimum Temperature :",min_temp)
print("Maximum Temperature :",max_temp)
```

```python
#Plot the graph
plt.plot(df["Timestamp"], df["Temperature"])
plt.xlabel("Time")
plt.ylabel("Temperature (°C)")
plt.title("Temperature Over Time")
plt.show()
```

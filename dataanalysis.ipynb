```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt 
```

```python
data = {
    "Name":["A","B","C","D","E"],
    "Math":[78,87,76,88,65],
    "Science":[67,89,64,81,95],
    "English":[69,80,90,79,91]
}
df = pd.DataFrame(data)
print(df)
```

```python
marks = df[["Math","Science","English"]].values
average_marks = np.mean(marks, axis =1)
df["Average"] = average_marks
print(df)
```

```python
#Let's assign the grade
def grade(avg):
    if avg >= 90:
        return 'A'
    elif avg >=75:
        return 'B'
    elif avg >=60:
        return 'C'
    else:
        return 'D' 

df['Grade']=df["Average"].apply(grade)
df
```

```python
# Let's plot the Graph
plt.figure()
plt.bar(df["Name"],df["Average"])
plt.xlabel("Student")
plt.ylabel("Average marks")
plt.title("Average marks of student")
plt.show()
```

```python
#average marks per subject
df[["Math","Science","English"]].mean().plot(kind="bar")
plt.title("average marks per subject")
plt.ylabel("Marks")
plt.show()
```
